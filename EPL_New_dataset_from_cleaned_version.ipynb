```
import pandas as pd

df = pd.read_csv('EPL_clean1.csv')#reading the new dataset same with EPL_4.csv
print(df.columns.tolist())#getting column names as a list
print(df.head(3))#observing first 3 observations
```

```
away = df[['Date','Season', 'away', 'away goals(full)','result(full)','away goals(half)', 'away shots',
           'away shots on target', 'away points','away corners','away yellow cards','away red cards','away fouls']].copy()#creating a new tabular dataset as away

print(away.head(3))
```

```
home = df[['Date','Season', 'home', 'home goals(full)','result(full)','home goals(half)', 'home shots',
           'home shots on target', 'home points','home corners','home yellow cards','home red cards','home fouls']].copy()#creating the tabular dataset called home

print(home.head(3))
```

```
home.columns=['date','season','team','goals(full)','result(full)','goals(half)','shots','shots on target','points','corners','yellow cards','red cards','fouls']#giving names to the home columns
print(home.head(3))
```

```
away.columns=['date','season','team','goals(full)','result(full)','goals(half)','shots','shots on target','points','corners','yellow cards','red cards','fouls']#giving names to the away columns
print(away.head(3))
```

```
#creating a venue column for both sides which tells us the team is whether home team or not
home['venue']='home'
away['venue']='away'
home['result(full)'] = home['result(full)'].map({'H':'W', 'A':'L', 'D':'D'})#mapping the result values( home win or away win) of home with win, draw or lose
away['result(full)'] = away['result(full)'].map({'A':'W', 'H':'L', 'D':'D'})#mapping  the result values(home win or away win) of away with win, draw or lose
print(home.head(2))
print(away.head(2))
```

```
team_df=pd.concat([home,away],ignore_index=True)#combining home and away tabular datasets
print(team_df.shape)#observing the amount of columns and rows
team_df.head(3)
```

```
team_df.to_csv('team_EPL3.csv',index=False)#saving the dataset
```
