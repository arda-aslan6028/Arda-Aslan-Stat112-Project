```python
#importing necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd


pd.set_option('display.max_columns', None) #displaying all columns



df = pd.read_csv('PremierLeague.csv',index_col=0) #reading the dataset and first observations
print(df.head())
df=pd.read_csv('PremierLeague.csv')
df.head()
```

```python
df.info()#getting information for columns with their data types and non-null value counts
```

```python
df.drop(df.columns[26:41], axis=1, inplace=True) #dropping unnecessary columns
df.drop(columns=['MatchID','MatchWeek','Referee','Time'],inplace=True)
df.info() #checking the remained columns
```

```python
df.head() #observing updated tabular data
```

```python
df = df[df['Season'] >= '2020-2021'].copy() #choosing the seasons between 2020-2021 and 2024-2025
df.head()
```

```python
df.isnull().sum() #checking whether there are null values
```

```python
#renaming columns into a proper name
df.rename(columns={'HomeTeam': 'home','AwayTeam': 'away','FullTimeHomeTeamGoals': 'home goals(full)','FullTimeAwayTeamGoals': 'away goals (full)','FullTimeResult': 'result(full)','HalfTimeHomeTeamGoals': 'home goals(half)','HalfTimeAwayTeamGoals': 'away goals(half)','HalfTimeResult': 'result(half)','HomeTeamShots': 'home shots','AwayTeamShots': 'away shots','HomeTeamShotsOnTarget': 'home shots on target','AwayTeamShotsOnTarget': 'away shots on target','HomeTeamCorners': 'home corners','AwayTeamCorners': 'away corners','HomeTeamFouls': 'home fouls','AwayTeamFouls': 'away fouls','HomeTeamYellowCards': 'home yellow cards','AwayTeamYellowCards': 'away yellow cards','HomeTeamRedCards': 'home red cards','AwayTeamRedCards': 'away red cards','HomeTeamPoints': 'home points','AwayTeamPoints': 'away points'},inplace=True)

df.info()
```

```python
#changing spsecific columns' data types
df['home goals(half)']=df['home goals(half)'].astype(int)
df['away goals(half)']=df['away goals(half)'].astype(int)
df['home shots']=df['home shots'].astype(int)
df['away shots']=df['away shots'].astype(int)
df['home shots on target']=df['home shots on target'].astype(int)
df['away shots on target']=df['away shots on target'].astype(int)
df['home corners']=df['home corners'].astype(int)
df['away corners']=df['away corners'].astype(int)
df['home fouls']=df['home fouls'].astype(int)
df['away fouls']=df['away fouls'].astype(int)
df['away yellow cards']=df['away yellow cards'].astype(int)
df['home yellow cards']=df['home yellow cards'].astype(int)
df['away red cards']=df['away red cards'].astype(int)
df['home red cards']=df['home red cards'].astype(int)
df.info()
```

```python

# fix the inconsistent column name
df = df.rename(columns={'away goals (full)': 'away goals(full)'})#fixing a naming mistake

# convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])#changing Date column's data type into Date
df.info()
df.head()
```

```python
df.to_csv('EPL_4.csv',index=False) #saving the new dataset
```
